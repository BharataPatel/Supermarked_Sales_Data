```python
import matplotlib.pyplot as plt
```

```python
cat_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))
plt.bar(cat_sales.index, cat_sales.values, color='purple')
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(8,5))
plt.bar(region_sales.index, region_sales.values, color='purple')
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
ship_count = df['Ship Mode'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(ship_count, labels=ship_count.index, autopct="%1.1f%%", colors=['purple','plum','orchid','mediumorchid'])
plt.title("Shipping Method Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = df["Category"].value_counts()

purple_shades = ['#800080', '#9932CC', '#8A2BE2', '#9370DB', '#BA55D3', '#DA70D6', '#D8BFD8', '#E6E6FA']

colors = [purple_shades[i % len(purple_shades)] for i in range(len(counts))]

plt.figure(figsize=(6,6))
plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    colors=colors
)

plt.title("Category Distribution (Purple Shades)")
plt.show()
```
