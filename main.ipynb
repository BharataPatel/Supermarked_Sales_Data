```python
configs = [
    ((1,1,1), (1,0,1,7)),
    ((2,1,1), (1,0,1,7)),
    ((1,1,2), (1,0,1,7)),
    ((1,0,1), (1,0,1,7)),
    ((1,1,1), (1,1,1,7))
]

for order, seasonal_order in configs:

    model = SARIMAX(
        train,
        order=order,
        seasonal_order=seasonal_order
    )

    fit = model.fit(disp=False)

    pred = fit.forecast(steps=len(validation))

    mae = mean_absolute_error(validation, pred)

    print(order, seasonal_order, "MAE:", mae)
```

```python
history = daily_sales.copy()
future_predictions = []

for date in future_dates:

    lag_1 = history.loc[date - pd.Timedelta(days=1)]
    lag_7 = history.loc[date - pd.Timedelta(days=7)]
    lag_14 = history.loc[date - pd.Timedelta(days=14)]
    lag_28 = history.loc[date - pd.Timedelta(days=28)]

    rolling_7 = history.loc[
        date - pd.Timedelta(days=7):
        date - pd.Timedelta(days=1)
    ].mean()

    rolling_28 = history.loc[
        date - pd.Timedelta(days=28):
        date - pd.Timedelta(days=1)
    ].mean()

    row = pd.DataFrame({
        "Lag_1": [lag_1],
        "Lag_7": [lag_7],
        "Lag_14": [lag_14],
        "Lag_28": [lag_28],
        "Rolling_7": [rolling_7],
        "Rolling_28": [rolling_28],
        "DayOfWeek": [date.dayofweek],
        "Month": [date.month]
    })

    prediction = final_model.predict(row)[0]

    future_predictions.append(prediction)

    history.loc[date] = prediction
```
